## Actividad 2. Funciones

Defina una función llamada `clasificar_nota(nota, corte=70)` que:

- reciba una nota;
- utilice `70` como corte predeterminado;
- retorne `"Aprobado"` si la nota es mayor o igual al corte;
- retorne `"Reprobado"` en caso contrario.

## Actividad 10. Valores faltantes

1. Identifique cuántos valores faltantes existen por columna.
2. Cree `df_limpio` como una copia de `df`.
3. Reemplace los faltantes de `horas_estudio` con la **mediana** de esa columna.
4. Reemplace los faltantes de `horas_sueno` con la **media** de esa columna.
5. Compruebe que esas dos columnas ya no tengan valores faltantes.


````python
print("Faltantes por columna:")
print(df.isna().sum())

df_limpio = df.copy()

df_limpio["horas_estudio"] = df_limpio["horas_estudio"].fillna(
    df_limpio["horas_estudio"].median()
)

df_limpio["horas_sueno"] = df_limpio["horas_sueno"].fillna(
    df_limpio["horas_sueno"].mean()
)

print("\nFaltantes después de limpiar:")
print(df_limpio[["horas_estudio", "horas_sueno"]].isna().sum())
````

Sobre `df_limpio`, obtenga:

2. Estudiantes de **Estadística** con asistencia mayor o igual a 90.
3. Estudiantes en modalidad **Híbrida** con nota entre 70 y 85, inclusive.
4. Estudiantes de **Economía o Actuariales** con nota mayor o igual a 80.

Recuerde utilizar paréntesis cuando combine condiciones.


````python
filtro_1 = df_limpio[df_limpio["nota"] >= 85]

filtro_2 = df_limpio[
    (df_limpio["carrera"] == "Estadística") &
    (df_limpio["asistencia"] >= 90)
]

filtro_3 = df_limpio[
    (df_limpio["modalidad"] == "Híbrida") &
    (df_limpio["nota"] >= 70) &
    (df_limpio["nota"] <= 85)
]

filtro_4 = df_limpio[
    ((df_limpio["carrera"] == "Economía") |
     (df_limpio["carrera"] == "Actuariales")) &
    (df_limpio["nota"] >= 80)
]

display(filtro_1)
display(filtro_2)
display(filtro_3)
display(filtro_4)
````

## Actividad 12. Crear y transformar variables

Agregue a `df_limpio`:

1. `nota_ajustada`: nota original + 3.
2. `estado`: `"Aprobado"` si `nota >= 70`; de lo contrario `"Reprobado"`, usando `np.where()`.
3. `estudio_alto`: `"Sí"` si `horas_estudio >= 7`; de lo contrario `"No"`, usando `np.where()`.
4. Convierta `nota` a tipo `float`.
5. Muestre las primeras cinco filas con las nuevas variables.


````python
df_limpio["nota_ajustada"] = df_limpio["nota"] + 3

df_limpio["estado"] = np.where(
    df_limpio["nota"] >= 70,
    "Aprobado",
    "Reprobado"
)

df_limpio["estudio_alto"] = np.where(
    df_limpio["horas_estudio"] >= 7,
    "Sí",
    "No"
)

df_limpio["nota"] = df_limpio["nota"].astype(float)

display(df_limpio.head())
````

## Actividad 13. Ordenamiento

1. Cree `ranking` ordenando de mayor a menor por `nota`.
2. Reinicie el índice de `ranking`.

````python
ranking = df_limpio.sort_values("nota", ascending=False)
ranking = ranking.reset_index(drop=True)

display(ranking[["id", "carrera", "nota"]].head())

orden_carrera = df_limpio.sort_values(
    ["carrera", "nota"],
    ascending=[True, False]
)

display(orden_carrera[["id", "carrera", "nota"]])
````

## Actividad 14. Agrupaciones

Construya `resumen_carrera` agrupando por `carrera` e incluyendo:

- promedio de `nota`;
- mediana de `nota`;
- nota máxima;
- promedio de `horas_estudio`;
- cantidad de estudiantes.

Luego construya un segundo resumen por `modalidad` con:

- promedio de `nota`;
- promedio de `asistencia`;
- cantidad de estudiantes.


````python
resumen_carrera = df_limpio.groupby("carrera").agg(
    promedio_nota=("nota", "mean"),
    mediana_nota=("nota", "median"),
    nota_maxima=("nota", "max"),
    promedio_horas_estudio=("horas_estudio", "mean"),
    cantidad=("id", "count")
)

resumen_modalidad = df_limpio.groupby("modalidad").agg(
    promedio_nota=("nota", "mean"),
    promedio_asistencia=("asistencia", "mean"),
    cantidad=("id", "count")
)

display(resumen_carrera)
display(resumen_modalidad)
````

## Actividad 15. Combinar tablas con `merge()`

Ejecute la tabla auxiliar:

```python
info_carreras = pd.DataFrame({
    "carrera": ["Estadística", "Economía", "Actuariales"],
    "area": ["Ciencias", "Ciencias Sociales", "Ciencias"],
    "codigo": ["EST", "ECO", "ACT"]
})
```

Observe que **Informática no aparece** en la tabla auxiliar.

Luego:

1. Haga un `merge()` con `how="inner"`.
2. Haga otro con `how="left"`.
3. Compare la cantidad de filas de ambos resultados.
4. Explique qué ocurrió con los registros de Informática.


````python
info_carreras = pd.DataFrame({
    "carrera": ["Estadística", "Economía", "Actuariales"],
    "area": ["Ciencias", "Ciencias Sociales", "Ciencias"],
    "codigo": ["EST", "ECO", "ACT"]
})

merge_inner = df_limpio.merge(
    info_carreras,
    on="carrera",
    how="inner"
)

merge_left = df_limpio.merge(
    info_carreras,
    on="carrera",
    how="left"
)

print("Filas inner:", merge_inner.shape[0])
print("Filas left:", merge_left.shape[0])

display(merge_inner.head())
display(merge_left[merge_left["carrera"] == "Informática"])
````

**Explicación:** el `inner` conserva únicamente las filas cuya carrera tiene coincidencia en ambas tablas, por lo que elimina los registros de Informática. El `left` conserva todas las filas de `df_limpio`; para Informática, las columnas `area` y `codigo` quedan con valores faltantes (`NaN`).
